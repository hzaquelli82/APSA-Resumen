```python
import MySQLdb
import pandas as pd

# Conectar a la base de datos MySQL local
try:
    db = MySQLdb.connect(
        host="127.0.0.1",
        user="operario",
        passwd="",
        db="dbp8100"
    )
    cur = db.cursor()
    print("Conexión exitosa a la base de datos dbp8100")
except MySQLdb.Error as e:
    print(f"Error al conectar a la base de datos: {e}")
```
